```python
import pandas as pd
import numpy as np
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
```

# Load Wisconsin Breast Cancer Dataset

```python
data = load_breast_cancer()
```

```python
y = 1 - data.target
```

# Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(data.data, y)
```

# Tuning AdaBoostClassifier

```python
ada = AdaBoostClassifier(base_estimator=DecisionTreeClassifier())
ada_params = {
    'base_estimator__max_depth': range(1,5),
    'n_estimators': [50, 100]
}
gs_ada = GridSearchCV(ada, param_grid=ada_params)
gs_ada.fit(X_train, y_train)
print(gs_ada.best_score_)
print(gs_ada.best_params_)
```

```python
gs_ada.score(X_test, y_test)
```

# GradientBoostingClassifier

```python
gb = GradientBoostingClassifier()
gb_params = {
    'max_depth': [2, 3],
    'n_estimators': [75, 100, 125],
    'max_features': ['auto', None]
}
gs_gb = GridSearchCV(gb, param_grid=gb_params)
gs_gb.fit(X_train, y_train)
print(gs_gb.best_score_)
print(gs_gb.best_params_)
```

```python
gs_gb.score(X_test, y_test)
```

# VotingClassifier

```python
vote = VotingClassifier([
    ('ada', AdaBoostClassifier(base_estimator=DecisionTreeClassifier())),
    ('gb', GradientBoostingClassifier())
])
vote_params = {
    'ada__n_estimators': [50, 100],
    'gb__n_estimators': [100, 125]
}
gs_vote = GridSearchCV(vote, param_grid=vote_params)
gs_vote.fit(X_train, y_train)
print(gs_vote.best_score_)
print(gs_vote.best_params_)
```

```python
gs_vote.score(X_test, y_test)
```
